```python
x1[1:4]
```

```python
x1[::2]
```

```python
x2[:, 1:]
```

```python
x2[0:2, 0:2]
```

```python
# 3D slicing: select the last row from every block
x3[:, 1:, :]
```

### 13. Common NumPy functions
- **`sum()`**: total; **`min()` / `max()`**: smallest/largest
- **`mean()`**: average; **`median()`**: middle value
- **`std()`**: standard deviation; **`round()`**: nearest values
- **`ceil()` / `floor()`**: round upward/downward

```python
values = np.array([2, 4, 6, 8])
np.sum(values)
```

```python
np.min(values), np.max(values)
```

```python
np.mean(values), np.median(values)
```

```python
np.std(values)
```

```python
decimals = np.array([1.2, 2.7, -1.5])
np.ceil(decimals)
```

```python
np.floor(decimals)
```

```python
np.round(decimals)
```

```python
matrix = np.array([[1, 2, 3], [4, 5, 6]])
np.sum(matrix, axis=0)  # Column totals
```

```python
np.sum(matrix, axis=1)  # Row totals
```

### 15. Stacking arrays
- **`hstack()`** joins arrays horizontally (more columns for 2D arrays).
- **`vstack()`** joins arrays vertically (more rows for 2D arrays).

```python
s1 = np.array([[1, 2], [3, 4]])
s2 = np.array([[5, 6], [7, 8]])
np.hstack((s1, s2))
```

```python
np.vstack((s1, s2))
```

```python
# Question 23: Stack two 1D arrays vertically.

# Write your answer below
stack_arr = np.vstack((s1,s2))
stack_arr

```
